```python
import numpy as np
import pandas as pd
from scipy.stats import t

datos = pd.read_csv(
    "../dataset/airfoil_self_noise.dat",
    sep=r"\s+",
    header=None,
    names=[
        "frequency", "angle", "chord_length",
        "velocity", "displacement", "sound_level"
    ]
)

x = datos["frequency"].to_numpy(dtype=float)
y = datos["sound_level"].to_numpy(dtype=float)

n = len(y)
x_media = x.mean()
y_media = y.mean()

# Ajustar la recta
Sxx = np.sum((x - x_media)**2)
Sxy = np.sum((x - x_media) * (y - y_media))

b1 = Sxy / Sxx
b0 = y_media - b1 * x_media

# Varianza estimada del error
y_estimado = b0 + b1 * x
SCE = np.sum((y - y_estimado)**2)
varianza_estimada = SCE / (n - 2)

# Error estándar de la pendiente
error_estandar_b1 = np.sqrt(varianza_estimada / Sxx)

# Estadístico de prueba bajo H0: beta_1 = 0
T = b1 / error_estandar_b1

# Valor crítico bilateral
alpha = 0.05
grados_libertad = n - 2
t_critico = t.ppf(1 - alpha / 2, grados_libertad)

# Valor p bilateral: complemento del contraste
valor_p = 2 * t.sf(abs(T), grados_libertad)

print(f"Pendiente estimada: {b1:.12f} dB/Hz")
print(f"Error estándar de la pendiente: {error_estandar_b1:.12f} dB/Hz")
print(f"Estadístico T: {T:.6f}")
print(f"Valor absoluto de T: {abs(T):.6f}")
print(f"Grados de libertad: {grados_libertad}")
print(f"t crítico: {t_critico:.6f}")
print(f"Valor p bilateral: {valor_p:.6e}")
```
