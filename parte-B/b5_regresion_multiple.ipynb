```python
# Copio y pego de a1_derivacion
import pandas as pd
import numpy as np

fuente = "../dataset/airfoil_self_noise.dat"
columnas = [
    "frequency",
    "angle",
    "chord_length",
    "velocity",
    "displacement",
    "sound_level"
]
df = pd.read_csv(fuente, sep="\t", names=columnas)
```

```python

# 1. Variables explicativas
variables_x = ["frequency", "angle", "chord_length", "displacement"]
X_datos = df[variables_x].to_numpy()

# 2. Variable de respuesta
Y = df["sound_level"].to_numpy()

n = len(Y)
k = 4

# 3. Construcción de la matriz X agregando la columna de unos para beta_0
columna_unos = np.ones((n, 1))
X = np.hstack([columna_unos, X_datos])

# 4. Resolución del sistema de ecuaciones normales en forma matricial:
# beta = (X^T * X)^(-1) * X^T * Y
Xt = X.T  # Matriz transpuesta de X
Xt_X = Xt @ X  # Producto matricial X^T * X
Xt_Y = Xt @ Y  # Producto matricial X^T * Y

# Inversa y estimación de los coeficientes beta
XtX_inv = np.linalg.inv(Xt_X)
beta = XtX_inv @ Xt_Y

beta_0, beta_1, beta_2, beta_3, beta_4 = beta
```

```python
# 6. Salida de resultados estructurada
print("=" * 60)
print("ESTIMACIÓN MATRICIAL DE MÍNIMOS CUADRADOS (B.5)")
print("=" * 60)
print(f"Número de observaciones (n): {n}")
print(f"Número de variables explicativas (k): {k}")
print("\nCoeficientes estimados:")
print(f"  beta_0 (Intercepto) : {beta_0:14.6f}")
print(f"  beta_1 (frequency)  : {beta_1:14.6f}")
print(f"  beta_2 (angle)      : {beta_2:14.6f}")
print(f"  beta_3 (chord_length): {beta_3:14.6f}")
print(f"  beta_4 (displacement): {beta_4:14.6f}")

print("\nEcuación del modelo final:")
print(
    f"  ŷ = {beta_0:.4f} + ({beta_1:.6f})*frequency + ({beta_2:.4f})*angle + "
    f"({beta_3:.4f})*chord_length + ({beta_4:.4f})*displacement"
)
```

```python
# 7. Predicciones del modelo múltiple
y_hat = (
    beta[0]
    + beta[1] * df["frequency"]
    + beta[2] * df["angle"]
    + beta[3] * df["chord_length"]
    + beta[4] * df["displacement"]
)

# 8. Residuos y suma de cuadrados del error (SCE)
residuos = Y - y_hat
SCE_multiple = np.sum(residuos**2)

# 9. R² y R² ajustado del modelo múltiple
Y_prom = np.mean(Y)
STC = np.sum((Y - Y_prom) ** 2)

R2_multiple = 1.0 - (SCE_multiple / STC)
R2_adj_multiple = 1.0 - (1.0 - R2_multiple) * ((n - k) / (n - k - 1))

print(f"R² múltiple: {R2_multiple:.4f}")
print(f"R² ajustado múltiple: {R2_adj_multiple:.4f}")
```
