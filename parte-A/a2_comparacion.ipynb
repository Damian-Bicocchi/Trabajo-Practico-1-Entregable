```python
# Copio y pego de a1_derivacion para tener los datos

import pandas as pd
import numpy as np

fuente = "../dataset/airfoil_self_noise.dat"
columnas = [
    "frequency",
    "angle",
    "chord_length",
    "velocity",
    "displacement",
    "sound_level"
]

df = pd.read_csv(
    fuente,
    sep="\t",
    header=None,
    names=columnas
)

print(df)

x = df["frequency"].values
y = df["sound_level"].values

```

```python
cantidad_elementos = len(x)

# Calculamos el promedio
x_media = np.mean(x)
y_media = np.mean(y)

# Calculamos Sxx y Sxy según la definición
Sxx = np.sum((x - x_media)**2)
Sxy = np.sum((x - x_media) * (y - y_media))

# Usamos las igualdades descubiertas en el informe
beta_1 = Sxy / Sxx
beta_0 = y_media - beta_1 * x_media

print("Cantidad de observaciones:", cantidad_elementos)
print("Media de x:", x_media)
print("Media de y:", y_media)
print("Sxx:", Sxx)
print("Sxy:", Sxy)

print("\nCoeficientes calculados mediante las fórmulas:")
print("Beta 0:", beta_0)
print("Beta 1:", beta_1)
```

```python
"""
Prompt: Podrías ahora generar un modelo lineal que permita modelar según Frecuencia y Sound level, guardandome los valores de beta_0 y beta_1 para poder compararlos

(vale aclarar, esto se hizo en la misma conversación, por lo que contaba con el contexto del prompt anterior) 
"""

from sklearn.linear_model import LinearRegression

# Crear el modelo
modelo = LinearRegression()

# Ajustar el modelo
# Se utiliza reshape(-1, 1) porque sklearn espera una matriz
# de dimensiones (n_observaciones, n_variables).
modelo.fit(x.reshape(-1, 1), y)

# Obtener los coeficientes
beta_0_sklearn = modelo.intercept_
beta_1_sklearn = modelo.coef_[0]

print("Coeficientes obtenidos con scikit-learn:")
print("Beta 0:", beta_0_sklearn)
print("Beta 1:", beta_1_sklearn)
```

```python
print("Comparación de resultados:")

print("-"*10)
print("Beta 0 según la fórmula:", beta_0)
print("Beta 0 según la recta:", beta_0_sklearn)
print("¿Coinciden?:", beta_0 == beta_0_sklearn)

print("-"*10)
print("Beta 1 según la fórmula:", beta_1)
print("Beta 1 según la recta:", beta_1_sklearn)
print("¿Coinciden?:", beta_1 == beta_1_sklearn)
```
